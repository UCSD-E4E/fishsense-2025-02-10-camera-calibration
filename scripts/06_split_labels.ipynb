```python
import json
```

```python
with open('../label_studio_results/project-35-at-2025-02-12-03-03-3fb00dba.json', 'r', encoding='utf-8') as handle:
    original_labels = json.load(handle)
```

```python
fsl_09d_tasks = [task for task in original_labels if task['data']['img'].find('09D') > 0]
fsl_11d_tasks = [task for task in original_labels if task['data']['img'].find('11D') > 0]
```

```python
with open('../label_studio_results/fsl-09d.json', 'w', encoding='utf-8') as handle:
    json.dump(fsl_09d_tasks, handle)
with open('../label_studio_results/fsl-11d.json', 'w', encoding='utf-8') as handle:
    json.dump(fsl_11d_tasks, handle)
```
